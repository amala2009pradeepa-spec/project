ABSTRACT
This project analyzes e-commerce sales data to identify top-selling products and monthly sales trends. The dataset was cleaned by removing duplicates and handling missing values. Using Pandas, NumPy, and Matplotlib, revenue analysis and data visualization were performed. Bar charts, line charts, and pie charts were used to present the results. The analysis helped identify the highest revenue-generating products and the best-performing months. This project demonstrates how data analytics can support business decision-making and improve sales performance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pandas as pd

data = {
    'Order_ID':[1001,1002,1003,1004,1005,1006,1007,1008,1009,1010,
                1011,1012,1013,1014,1015,1016,1017,1018,1019,1020],

    'Order_Date':['2025-01-05','2025-01-10','2025-02-12','2025-02-18',
                  '2025-03-08','2025-03-15','2025-04-05','2025-04-20',
                  '2025-05-12','2025-05-28','2025-06-10','2025-06-25',
                  '2025-07-07','2025-07-18','2025-08-03','2025-08-16',
                  '2025-09-05','2025-09-22','2025-10-08','2025-10-20'],

    'Product':['Laptop','Mouse','Keyboard','Laptop','Monitor',
               'Mouse','Keyboard','Laptop','Monitor','Mouse',
               'Laptop','Keyboard','Monitor','Mouse','Laptop',
               'Keyboard','Monitor','Mouse','Laptop','Keyboard'],

    'Quantity':[2,5,3,1,2,10,4,3,1,8,2,5,3,12,1,6,2,7,2,4],

    'Price':[50000,500,1200,50000,15000,500,1200,50000,15000,500,
             50000,1200,15000,500,50000,1200,15000,500,50000,1200]
}

df = pd.DataFrame(data)

print(df.head())
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Missing Values Check
print(df.isnull().sum())

# Remove Duplicates
df.drop_duplicates(inplace=True)

# Date Conversion
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

# Revenue Column
df['Revenue'] = df['Quantity'] * df['Price']

print(df.head())
```

```python
top_products = df.groupby('Product')['Revenue'].sum().sort_values(ascending=False)

print(top_products)
```

```python
plt.figure(figsize=(8,5))
top_products.plot(kind='bar')

plt.title('Top Selling Products')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.show()
```

```python
df['Month'] = df['Order_Date'].dt.month_name()

monthly_sales = df.groupby('Month')['Revenue'].sum()

print(monthly_sales)
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index,
         monthly_sales.values,
         marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid(True)

plt.show()
```

```python
# Revenue by Product
product_revenue = df.groupby('Product')['Revenue'].sum()

plt.figure(figsize=(7,7))

plt.pie(
    product_revenue,
    labels=product_revenue.index,
    autopct='%1.1f%%'
)

plt.title('Revenue Distribution by Product')
plt.show()
```
